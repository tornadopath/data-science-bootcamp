```python
price = pd.Series(data={'Apple': 200, 'intellia':12, 'Crispr':60, 'KGHM':np.nan})
price
```

```python
price['intellia'] #lub index
```

```python
price.count() #zliczanie obiektów
```

```python
price.value_counts(dropna=False) ### WAŻNE, LICZY ILE RAZY COŚ WYSTAPIŁO, dropna - liczy rowniez brak danych
```

```python
price.sum()
```

```python
price.max()
```

```python
price.describe()
```

```python
price.nlargest(1) #najwyższa wartość
```

```python
price.nsmallest(2)
```

```python
price.rank() - #uporzadkowanie
```

```python
price.sort_values(ascending=False) #sortowanie
```
